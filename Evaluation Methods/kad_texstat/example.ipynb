```python
from texstat.kad import KAD_wrapper

# TexStat model parameters
sr            = 44100
frame_size    = sr*2
hop_size      = frame_size//2
bandwidth     = None
kernel        = 'gaussian'
device        = 'cuda'
class_1 = "/home/esteban/Documents/projects/dev_kad/data/Bees_orig"
class_2 = "/home/esteban/Documents/projects/dev_kad/data/Bees_gen"

KAD_texstat = KAD_wrapper(
    frame_size = frame_size, 
    sampling_rate = sr,
    hop_size = hop_size,  # No overlap (can set to frame_size//2 for 50% overlap)
    bandwidth = bandwidth, # bandwidths = [0.1, 1.0, 10.0, None]  # None = automatic
    kernel = kernel, # kernels = ['gaussian', 'iq', 'imq']
    device=device
)

# Compute KAD score using the full set of statistics
kad_score = KAD_texstat.score(class_1, class_2, cache_dir='embeddings_sub', model="full")
print(f"KAD score: {kad_score:.4f}")
```
